```python
#Moon phase calculator

A python script that calculates the moons phase and percentage of illumination for any given date
**How it works;**
Uses a known new moon date alongside with the moons - 29.53 days synodic period to determine the progression of any given date
The calculated progression determines the moons phase and percentage of illumination
**My learnings**
-Working with datetime feature
-Writing and using functions
-Orbital period calculations and applying it in the code
-Testing against a known value


from datetime import datetime
import math
#-----Step1:- the building blocks we need

# A date we know for certain was a "New Moon"
reference_new_moon = datetime(2000 , 1 , 6 , 18 , 14)

# the moon takes about 29.53 days to go through a full cycle
synodic_month = 29.530588861

# The phase names , and where they start in the cycle ( 0.0 to 1.0)
phase_names = [
    (0.0 , "New Moon") ,
    (0.03 , "Waxing Crescent") ,
    (0.22 , "First Quarter") ,
    (0.28 , "Waxing Gibbous") ,
    (0.47 , "Full Moon") ,
    (0.53 , "Waning Gibbous") ,
    (0.72 , "Last Quarter") ,
    (0.78 , "Waning Crescent") ,
    (0.97 , "New Moon"),
]

#-----Step 2: The function  -----
def get_moon_phase(date):
  #How many days have passed since our new moon?
  days_passed = (date-reference_new_moon).total_seconds()/86400

  #Where are we in the Current cycle ? (a number between 0 and 29.53)
  position_in_cycle = days_passed % synodic_month

  #Turn the into a fraction between 0 and 1
  fraction = position_in_cycle / synodic_month

  #Figure out the phase name by checking each threshold
  phase_name= phase_names[0][1]
  for threshold, name in phase_names:
    if fraction >= threshold:
      phase_name = name
      #Estimate how much of the moon is lit up(0% to 100%)
 illumination = (1 - math.cos(2*math.pi*fraction))
 return phase_name, illumination
#-------Step 3: Use the function -----
today = datetime.now()
name,illum= get_moon_phase(today)
print(f"Today's moon phase: {name} ({illum:.1f}% illuminated)")

```
